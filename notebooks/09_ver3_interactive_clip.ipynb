## Reading the result

Selections are 30 seconds under the default checkpoint settings. A song shorter than 30 seconds is right-padded for the model; playback contains only real audio. Moving the slider clears stale predictions but does not run the model.

The entire song is held in RAM only for responsive playback/seeking. Unlike Version 2, no full-recording analysis is required before prediction. Features depend solely on the selected passage. Vocals and accompaniment remain together; this is not source separation.

The native STFT is larger, so inference defaults to one window per batch. Scores compare seven supported classes and are not calibrated probabilities. Install the unified requirements and restart JupyterLab if widget controls do not render.
